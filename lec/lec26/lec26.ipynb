## Set Up the Notebook

```python
from datascience import *
import numpy as np
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
import scipy
import plotly.express as px
from ipywidgets import interact, interactive, fixed, interact_manual
import ipywidgets as widgets
import warnings
warnings.simplefilter(action='ignore', category=UserWarning)

def graph_norm(mu=0, sigma=1):
    x=np.arange(-10, 10, 0.01)
    display(px.line(x=x, 
                    y=scipy.stats.norm.pdf(x, loc=mu, scale=sigma), 
                    range_y=[0, 1]));

def graph_norm_area(mu=0, sigma=1, SD_range=0):
    x = np.arange(-3, 3, 0.01)
    y = scipy.stats.norm.pdf(x, loc=mu, scale=sigma)
    
    lower = mu - SD_range * sigma
    upper = mu + SD_range * sigma
    
    prob = scipy.stats.norm.cdf(upper, loc=mu, scale=sigma) - \
           scipy.stats.norm.cdf(lower, loc=mu, scale=sigma)
    
    label = f'±{SD_range:.1f}SD = {prob*100:.2f}%'
    
    df = Table().with_columns(
        'x', x,
        'y', y,
        'region', np.where((x >= lower) & (x <= upper), label, 'outside')
    )
    
    fig = px.area(df.to_df(), x='x', y='y', color='region',
                  color_discrete_map={label: 'steelblue', 'outside': 'lightgrey'},
                  range_y=[0, 1])
    
    # Add percentage annotation over the shaded region
    fig.add_annotation(
        x=mu, 
        y=scipy.stats.norm.pdf(mu, loc=mu, scale=sigma) / 2,
        text=f'{prob*100:.2f}%',
        showarrow=False,
        font=dict(size=14, color='steelblue')
    )
    
    display(fig)
```

## Standard Units Review

* How many SDs above average?
* `z = (value - average)/SD`
    * Negative z: value below average
    * Positive z: value above average
    * z = 0: value equal to average
* When values are in standard units: average = 0, SD = 1
* Gives us a way to compare/understand data no matter what the original units

### Probability Density Curve

The height of the (probability density) curves are determined by the formula

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2}$$

and the shape is determined by the mean (`mu`, $\mu$) and standard deviation (`sigma`, $\sigma$) values.

_Note: You will not be working with this formula directly in this class, but we can use a python representation of it (`scipy.stats.norm.pdf`) to see how the mean and standard deviation impact the shape of the distribution._

### How Big are Most of the Values?

* No matter what the shape of the distribution, the bulk of the data are in the range "average ± a few SDs" (Chebyshev's Inequality)
* **If a histogram is bell-shaped**, then almost all of the data are in the range "average ± 3 SDs"

### A "Central" Area

The approximate percentages below refer to the area under the Normal curve, above the horizontal axis, and within 1, 2, and 3 standard deviations of the mean:

* ~68% of data falls within ±1 standard deviations of the mean
* ~95% of data falls within ±2 standard deviations of the mean
* ~99.7% of data falls within ±3 standard deviations of the mean

These percentages are calculated using the CDF. Slide the `SD_range` slider to explore the relationship between the area and the number of standard deviations from the mean.

```python
interact(graph_norm_area,
         mu=(-1, 1),
         sigma=(0.2**0.5, 5**0.5, 0.05),
         SD_range=(0.0, 4.0, 0.1));
```
